## Data Aggregation and Grouping

```python
s1 = pd.Series(data=['a','b','a','b','c','c'])
s2 = pd.Series(data=[1,2,1,2,5,5])
data = pd.DataFrame({
    "name":s1,
    "lemonades":s2
})
data
```

```python
dataGroup = data.groupby("name")
dataGroup.sum()
```

```python
dataGroup["lemonades"].mean()
```

```python
dataGroup["lemonades"].max()
```

```python
## by Default pivot table give average ... can use aggfunc for other aggregration 
```

```python
data.pivot_table(values="lemonades",index="name")
```

```python
## cross tabs
pd.crosstab(data['name'],data['lemonades'])
```

## Merging Joining and Concatination

```python
s1 = pd.Series(data=['a','b','a','b','c'])
s2 = pd.Series(data=[1,2,1,2,5])
data1 = pd.DataFrame({
    "name":s1,
    "lemonades":s2
})

s1 = pd.Series(data=['a','a','a','b','c'])
s2 = pd.Series(data=[10,20,10,20,50])
data2 = pd.DataFrame({
    "name":s1,
    "lemon":s2
})
```

```python
pd.concat([data1,data2])
```

```python
pd.merge(data1,data2,on='name')
```

```python
data1.join(data2.set_index('name'), on='name')
```
