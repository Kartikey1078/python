```python
import matplotlib.pyplot as plt
## pylot is collection of function that make matplotlib work as matlab
```

## create chart 

```python
x = [1,2,3,4,5]
y = [20,30,40,60,70]
plt.plot(x,y) ## line chart 
plt.show() ## dislay chart 
```

## customise chart 

```python
plt.figure(figsize=(6,4)) ## figure size 

plt.plot(x,y,color='green',marker='o',linestyle='dashed',linewidth=2,markersize=20)

plt.title("lable yha se aaega")
plt.xlabel("x axis ka label yha se aaega")
plt.ylabel("y axis ka label yha se aaega")
plt.show()
```

## Multi data line chart 

```python
x = [1,2,3,4,5]
y = [20,30,40,60,70]
y2 = [200,320,420,620,720]

plt.plot(x,y,label='2024 data')
plt.plot(x,y2,label='2022 data')

plt.title("diffrent year data")
plt.xlabel("data of both year")
plt.ylabel("data of both year")
plt.legend()

plt.show()

```

## Bar chart 

```python
x = [1,2,3,4,5]
y = [20,30,40,60,70]
y2 = [200,320,420,620,720]

plt.bar(x,y)
plt.title("bar chart")
plt.xlabel("x label")
plt.ylabel("y label")

plt.show()
```

## histogram

```python
## distribution analysis
## histogram show continous data vs bar chart show only catefogry data 
import random 
data = [random.randint(1,100) for _ in range(500)]
plt.hist(data,bins=20)
plt.title("histogram title")
plt.show()
```

## pie chart 

```python
categories = ['a','b','c','d','e']
sales = [20,30,40,60,70]

plt.pie(sales,labels=categories,autopct='%1.2f%%',startangle=90)
plt.title("title h ye")

plt.show()
```

## scatter plot

```python
sales = [20,30,40,60,70]
sales2 = [120,130,140,160,170]

plt.scatter(sales,sales2)
```

## sub plots

```python
## use to show multiple chart in one figure

##bar chart##
sales = ['m','t','w','t','f','s','s']
sales2 = [120,130,140,160,170,11,67]
## scatter plot chart 
data1 = [20,30,40,60,70]
data2 = [120,130,140,160,170]

plt.figure(figsize=(15,8))

plt.subplot(1,2,2)## row,column,position
plt.bar(sales,sales2)
plt.title("bar chart sales")
plt.xlabel("x lable")
plt.ylabel("y label")

plt.subplot(1,2,1)## row,column,position
plt.scatter(data1,data2)
plt.title("scatter chart sales")
plt.xlabel("x lable")
plt.ylabel("y label")

plt.show()
```
